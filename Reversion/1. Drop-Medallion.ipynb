```sql
%python
dbutils.widgets.removeAll()
```

```sql
create widget text storageName default "adlssmartprojectdev13";
create widget text catalogo default "project_dev";
create widget text schemaBronze default "bronze";
create widget text schemaSilver default "silver";
create widget text schemaGold default "gold";
```

```sql
%python

storageName = dbutils.widgets.get("storageName")
schemaBronze = dbutils.widgets.get("schemaBronze")
schemaSilver = dbutils.widgets.get("schemaSilver")
schemaGold = dbutils.widgets.get("schemaGold")
catalogo = dbutils.widgets.get("catalogo")
```

```sql
%python

ruta_bronze = f"abfss://{storageName}@{schemaBronze}.dfs.core.windows.net"
ruta_silver = f"abfss://{storageName}@{schemaSilver}.dfs.core.windows.net"
ruta_gold = f"abfss://{storageName}@{schemaGold}.dfs.core.windows.net"
```

### Eliminacion tablas Bronze

```sql
-- DROP TABLES
DROP TABLE IF EXISTS ${catalogo}.${schemaBronze}.orders_item;
DROP TABLE IF EXISTS ${catalogo}.${schemaBronze}.orders_payments;
DROP TABLE IF EXISTS ${catalogo}.${schemaBronze}.orders_review;
DROP TABLE IF EXISTS ${catalogo}.${schemaBronze}.orders;
DROP TABLE IF EXISTS ${catalogo}.${schemaBronze}.products;
DROP TABLE IF EXISTS ${catalogo}.${schemaBronze}.products_category;
DROP TABLE IF EXISTS ${catalogo}.${schemaBronze}.customers;
DROP TABLE IF EXISTS ${catalogo}.${schemaBronze}.sellers;
DROP TABLE IF EXISTS ${catalogo}.${schemaBronze}.geolocation;
```

```sql
%python
## REMOVE DATA (Bronze)
dbutils.fs.rm(f"{ruta_bronze}/orders_item", True)
dbutils.fs.rm(f"{ruta_bronze}/orders_payments", True)
dbutils.fs.rm(f"{ruta_bronze}/orders_review", True)
dbutils.fs.rm(f"{ruta_bronze}/orders", True)
dbutils.fs.rm(f"{ruta_bronze}/products", True)
dbutils.fs.rm(f"{ruta_bronze}/products_category", True)
dbutils.fs.rm(f"{ruta_bronze}/customers", True)
dbutils.fs.rm(f"{ruta_bronze}/sellers", True)
dbutils.fs.rm(f"{ruta_bronze}/geolocation", True)
```

### Eliminacion tablas Silver

```sql
-- DROP TABLES
DROP TABLE IF EXISTS ${catalogo}.${schemaSilver}.orders_transform;
DROP TABLE IF EXISTS ${catalogo}.${schemaSilver}.customers;
DROP TABLE IF EXISTS ${catalogo}.${schemaSilver}.products_transform;
DROP TABLE IF EXISTS ${catalogo}.${schemaSilver}.sellers;
DROP TABLE IF EXISTS ${catalogo}.${schemaSilver}.orders_transform_end;
```

```sql
%python
# REMOVE DATA (Silver)
dbutils.fs.rm(f"{ruta_silver}/orders_transform", True)
dbutils.fs.rm(f"{ruta_silver}/customers", True)
dbutils.fs.rm(f"{ruta_silver}/products_transform", True)
dbutils.fs.rm(f"{ruta_silver}/sellers", True)
dbutils.fs.rm(f"{ruta_silver}/orders_transform_end", True)
```

### Eliminacion tablas Golden

```sql
-- DROP TABLES
DROP TABLE IF EXISTS ${catalogo}.${schemaGold}.orders;
```

```sql
%python
# REMOVE DATA (Golden)
dbutils.fs.rm(f"{ruta_gold}/orders", True)
```
